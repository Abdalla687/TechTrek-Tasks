```python
selection_to_dis = df.groupby('Meal')[['Channel','Rating']]
selection_to_dis.head(5)
```

```python
df['revenue'] = df['Qty'] * df['Price']
df
```

```python
total_total_revenue = df.groupby('revenue')[['Category','Branch']].sum()
total_total_revenue
```

```python
compare_avg = df.groupby('Channel')['Rating'].mean()
compare_avg
```

```python
hero_meal = selection_to_dis.idxmax()
hero_meal
```

```python
best_channel = compare_avg.idxmax()
best_channel
```

```python
best_branch = df.groupby("Branch")["revenue"].sum()

print(best_branch)
```

```python
best_branch_to_watch = best_branch.idxmax()
best_branch_to_watch
```
