```python
df.isnull().sum()
```

```python

df['Rating']=df['Rating'].fillna(df['Rating'].mean())
df
```

```python
df['progress_level']=''
for i in range(len(df)):
    if df.loc[i,'Score']>85:
        df.loc[i,'progress_level'] = 'Excellent'
    elif df.loc[i,'Score']<65:
        df.loc[i,'progress_level'] = 'Low'
    else:
        df.loc[i,'progress_level'] = 'Good'

df
```

```python
df['Is_Completed'] = df['Done'] == 'Yes'

summarize_df = df.groupby('Course').agg(
    Avg_Score=('Score', 'mean'), Completed_Count=('Is_Completed', 'sum')
)

summarize_df
```

```python
best_course = summarize_df.idxmax()
best_course
```
