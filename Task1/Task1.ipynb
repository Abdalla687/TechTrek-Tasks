```python
# Task 1
# Fill an array of 5 elements from the user, Sort it in descending and ascending orders then display the output.

a = []
for i in range(5):
  elements = int(input(f"Enter Any numbers to add in array {i+1}:"))
  a.append(elements)
print('List:',a)
a.sort()
print('Ascending Order:',a)
a.reverse()
print('Descending Order:',a)

```

```python
# Task 1
# Write a program that remove all vowels from the input word and generate a brief version of it.
vowels = "aeiouAEIOU"
text = input("Enter Any Text:")
for c in text:
  if c in vowels:
    replace = text.replace(c," ")
print (replace)
```

```python
# Task 1
# Write a program that prints the locations of "i" character in any string you added.

str1 = input("Enter Any Text :")
char = "i"
for i in str1:
  if i == char:
    location = str1.index(char)
print("The Location is:",location)
```
