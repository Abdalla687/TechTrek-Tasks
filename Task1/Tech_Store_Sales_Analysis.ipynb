product to promote.

```python
electronic_order = df[(df['Category']=='Electronics')&(df['Profit']>100)]
electronic_order
```

```python
label_based = df.loc[0:2,['Product','UnitPrice']]
label_based 
```

```python
pos_label = df.iloc[0:2,[3,4]]
pos_label
```

```python
df['Gross Sales']=df['UnitPrice'] * df['Quantity']
df['DiscountAmount'] = df['Gross Sales'] * df['Discount']
df['NetSales'] = df['Gross Sales'] - df['DiscountAmount']
df['Profit Margin'] = (df['Profit']/df['UnitPrice'])*100

cols_to_show = ['OrderID', 'Product', 'Gross Sales', 'DiscountAmount', 'NetSales', 'Profit', 'Profit Margin']
print(df[cols_to_show])

```

```python
summarize_cat_region = df.groupby(['Category','Region'])[['NetSales','Profit']].sum()
summarize_cat_region.sort_values(by='Profit',ascending=False)
```

```python
print("Best Category :\n",df.groupby('Category')['NetSales'].sum().sort_values(ascending = False))
```

```python
print("Strongest Region :\n",df.groupby('Region')['NetSales'].sum().sort_values(ascending = False))
```

```python
print("One Product to promote based on Net Sales \n",df.groupby('Product')['NetSales'].sum().sort_values(ascending=False))
```
