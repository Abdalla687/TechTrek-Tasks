```python
selected_cols = ['Name','Salary','PerformanceScore']
df[selected_cols].head(4)
```

```python
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
df['Salary'].isnull().sum()
```

```python
filtered_exp_max = df['ExperienceYears'].max()
filtered_exp_min = df['ExperienceYears'].min()
filtered_exp_avg = df['ExperienceYears'].mean()

print("filtered_exp_max : \n",filtered_exp_max)
print("filtered_exp_min : \n",filtered_exp_min)
print("filtered_exp_avg : \n",filtered_exp_avg)
```

```python
df['Employee Level']=''

for i in range(len(df)):
    if df.loc[i,'ExperienceYears']>4:
        df.loc[i,'Employee Level'] = 'Junior'
    elif df.loc[i,'ExperienceYears']==4:
        df.loc[i,'Employee Level'] = 'Mid'
    else:
        df.loc[i, 'Employee Level'] = 'Senior'

df.head(5)
```

```python
summarize_cols = df.groupby('Department')[['Salary','PerformanceScore']].mean()
summarize_cols

```
